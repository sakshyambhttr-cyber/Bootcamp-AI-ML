# Array Operations

Arithmetic Operations

```python
a1 = np.array([1,2,3,4,5])
```

```python
a2 = np.array([6,7,8,9,10])
```

```python
a1+a2
```

Whenever there is diff size of arrays there will be error in addition operation

```python
a1-a2
```

```python
a1*a2
```

```python
a1/a2
```

```python
a1 // a2
```

```python
a1**a2
```

Boardcasting

```python
x = [10,20,30,40]
```

```python
arr = np.array(x)
```

```python
arr+10
```

Deep and Shallow Copy


```python
a = np.arange(1,21)
```

```python
a
```

```python
slice = a[:5]
```

```python
slice
```

```python
slice = slice *10
```

Matrix Operations

```python
mat1 = np.array([[1,2],[3,4]])
```

```python
mat2 = np.array([[5,6],[7,8]])
```

```python
mat1*mat2
```

```python
mat1 @ mat2
```

```python
np.dot(mat1,mat2)
```

Stacking Array

```python
np.vstack((mat1,mat2))
```

```python
np.hstack((mat1,mat2))
```

```python
np.column_stack((mat1,mat2))
```

```python
c = np.arange(16).reshape(4,4)
c
```

```python
np.hsplit(c,4)
```

```python
np.vsplit(c,2)
```
