### 1. Import NumPy


```python
import numpy as np
print(f"NumPy Version: {np.__version__}")
```

### 2. Array Creation & Initialization
Creating 1D, 2D arrays, ranges, and special matrices.


```python
# Basic array creation
array_1d = np.array([1, 2, 3, 4, 5])
print("1D Array:", array_1d)

# Arange and Linspace
range_vec = np.arange(1, 10)
print("Arange 1..9:", range_vec)

linspace_arr = np.linspace(0, 1, 5)
print("Linspace 0..1 (5 steps):", linspace_arr)

# Zeros, Ones, and Eye matrices
zero_mat = np.zeros((3, 3))
print("Zeros Matrix:\n", zero_mat)

ones_mat = np.ones((2, 4))
print("Ones Matrix:\n", ones_mat)

identity_mat = np.eye(4)
print("Identity Matrix:\n", identity_mat)
```

### 3. Random Array Generation
Generating uniform and integer random arrays.


```python
# Seed for reproducibility
np.random.seed(42)

rand_uniform = np.random.rand(3, 3)
print("Random Uniform (3x3):\n", rand_uniform)

rand_integers = np.random.randint(1, 100, size=(4, 4))
print("Random Integers (4x4):\n", rand_integers)
```

### 4. Indexing, Slicing, and Reshaping
Manipulating array dimensions, slicing rows/columns, and reshaping.


```python
# 1D Indexing and Slicing
num_sequence = np.arange(11, 21)
print("Full Sequence:", num_sequence)
print("Element at index 3:", num_sequence[3])
print("Slice [2:7]:", num_sequence[2:7])
print("Slice with step [1:8:2]:", num_sequence[1:8:2])

# Reshaping 1D to 2D
grid_matrix = np.arange(1, 31).reshape(6, 5)
print("Reshaped 6x5 Matrix:\n", grid_matrix)

# 2D Slicing
sub_grid = grid_matrix[1:4, 2:5]
print("Sub-matrix slice (rows 1:4, cols 2:5):\n", sub_grid)
```

### 5. Boolean Masking and Filtering
Filtering array elements based on conditions.


```python
sample_arr = np.array([12, 45, 67, 23, 89, 34, 90, 15])
mask_condition = sample_arr > 30

print("Original Array:", sample_arr)
print("Boolean Mask (>30):", mask_condition)
print("Filtered Values:", sample_arr[mask_condition])
```

### 6. Elementwise Arithmetic Operations
Performing basic mathematical operations on NumPy arrays.


```python
vec_a = np.array([1, 2, 3, 4, 5])
vec_b = np.array([10, 20, 30, 40, 50])

print("Addition:", vec_a + vec_b)
print("Subtraction:", vec_b - vec_a)
print("Multiplication:", vec_a * vec_b)
print("Division:", vec_b / vec_a)
print("Exponentiation:", vec_a ** 2)
```

### 7. Matrix Operations & Transposition
Matrix multiplication, dot product, and transpose.


```python
mat_x = np.array([[1, 2], [3, 4]])
mat_y = np.array([[5, 6], [7, 8]])

# Matrix Dot Product
dot_product_res = np.dot(mat_x, mat_y)
print("Dot Product:\n", dot_product_res)

# Matrix Transpose
print("Transpose of X:\n", mat_x.T)
```

### 8. Statistical and Aggregate Functions
Computing sum, mean, std, min, max, and argmax.


```python
data_matrix = np.array([[10, 20, 30], [40, 50, 60], [70, 80, 90]])

print("Data Matrix:\n", data_matrix)
print("Total Sum:", np.sum(data_matrix))
print("Column-wise Sum (axis=0):", np.sum(data_matrix, axis=0))
print("Row-wise Sum (axis=1):", np.sum(data_matrix, axis=1))

print("Overall Mean:", np.mean(data_matrix))
print("Standard Deviation:", np.std(data_matrix))
print("Min Value:", np.min(data_matrix))
print("Max Value:", np.max(data_matrix))
print("Index of Max Value:", np.argmax(data_matrix))
```

### 9. Universal Functions (ufuncs)
Applying math, exponential, and trigonometric functions.


```python
val_arr = np.array([1, 4, 9, 16, 25])

print("Square Root:", np.sqrt(val_arr))
print("Exponential:", np.exp(val_arr[:3]))
print("Logarithm:", np.log(val_arr))
print("Sine Values:", np.sin(val_arr))
```
